### Column: seasons

- Most missing values are in seasons column because the title is a Movie.

- A movie has no seasons.
- These are logical missing values, not data errors.
Do not fill or remove these values.

````python
# Visualizing the missing values
df1.replace(["null", "NULL", "NaN", "None"], np.nan, inplace=True)
````

````python
df1.info()
````

````python
df1.describe().T
````

### Column : **seasons**

Datatype : 'float64'

````python
df1["seasons"].skew() # data is extremely positively (right) skewed.
````

A positive skew means:

- Most values are small.
- A few values are very large.
- The distribution has a long tail to the right.

````python
df1["seasons"].median()
````

````python
# Impute median in place of missing values in column seasons

df1["seasons"].fillna(df1["seasons"].median(),inplace=True)
````

````python
df1.isnull().sum().sort_values(ascending=False)
````

### Column : **age_certification**

Datatype : 'object'

````python
df1["age_certification"].mode() # find most frequent value
````

````python
# Impute R in place of missing values in column age_certification

df1["age_certification"].fillna("R",inplace=True)
````

````python
df1.isnull().sum().sort_values(ascending=False)
````

### Column : **character**

Datatype : 'object'

````python
df1['character'].mode() # find most frequent value
````

````python
# Impute himself in place of missing values in column character

df1['character'].fillna('Himself', inplace=True)
````

````python
df1.isnull().sum().sort_values(ascending=False)
````

### Column : **tmdb_score**

Datatype : 'float64'

````python
df1['tmdb_score'].skew() # Approximately symmetric
````

````python
plt.figure(figsize=(6,4))
sns.histplot(df1,x='tmdb_score',bins=20,kde=True)
plt.show()
````

````python

df1['tmdb_score'].mean()
````

````python
# The histogram chart shows that the data is distributed evenly.
# Fill missing values of tmdb_score column with mean.
df1['tmdb_score'].fillna(df1['tmdb_score'].mean(), inplace=True)
````

````python
df1.isnull().sum().sort_values(ascending=False)
````

### Column : **imdb_votes**

Datatype : 'float64'

````python
df1['imdb_votes'].skew() # Highly positively (right) skewed.
````

````python
# Fill with median
df1['imdb_votes'].fillna(df1['imdb_votes'].median(), inplace=True)
````

````python
df1.isnull().sum().sort_values(ascending=False)
````

### Column : **imdb_score**

Datatype : 'float64'

````python
df1["imdb_score"].skew() # Approximately symmetric
````

````python
plt.figure(figsize=(5,3))
sns.histplot(df1,x='imdb_score',bins=20,kde=True)
plt.show()
````

````python
# Replace missing 'imdb_score' values with the mean
df1['imdb_score'].fillna(df1['imdb_score'].mean(), inplace=True)
````

````python
df1.isnull().sum().sort_values(ascending=False)
````

### Column : **imdb_id**

Datatype : 'object'

````python
# imdb_id are unique values we can not impute them with mode
# Replace missing 'imdb_id' values with unknown

df1['imdb_id'].fillna('unknown', inplace=True)
````

````python
df1.isnull().sum().sort_values(ascending=False)
````

### Column : **description**

### Column : **tmdb_popularity**

````python
df1.isnull().sum().sort_values(ascending=False)
````

### Insights:
- The dataset initially contained 168 duplicate entries, which were removed using the drop_duplicates() method.

- Several columns had missing values, which were handled as follows :

 - 'tmdb_popularity' and 'description' columns had missing values and were dropped from the dataset.

 - Categorical columns such as 'age_certification' and 'character' (object types) had missing values that were filled using the mode.

 - Numerical columns like 'seasons' and 'imdb_votes', which were skewed, had their missing values filled using the median .

 - For columns like 'imdb_score' and 'tmdb_score', which followed a normal distribution, missing values were imputed using the mean.

# Exploratory Data Analysis (EDA)

## Univariate Analysis

In Univariate Analysis, we analyze one variable at a time.

The purpose is to understand the distribution, frequency, and characteristics of individual variables.

## Chart 1. Distribution of Movies and TV Shows

````python
# Chart 1: Distribution of Movies and TV Shows

plt.figure(figsize=(7,4))

ax = sns.countplot(
    data=df1,
    x='type',
    hue='type',
    palette='Set2',
    legend=False
)

# Display count labels on each bar
for container in ax.containers:
    ax.bar_label(container, fontsize=11)

plt.title('Distribution of Movies and TV Shows', fontsize=15, fontweight='bold')
plt.xlabel('Content Type', fontsize=10, fontweight='bold')
plt.ylabel('Number of Titles', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
````

### Why did I choose this chart?

A Count Plot is the most suitable visualization for categorical variables because it compares the frequency of each category. Since the type column contains only MOVIE and SHOW, this chart provides a clear comparison of the platform's content distribution.



### Insights
- Movies constitute the majority of Amazon Prime Video's content library.
- TV Shows represent a smaller portion of the catalog.
- The platform appears to focus more heavily on movie content.

### Business Impact

- Understanding the distribution of content helps Amazon evaluate whether its library is balanced. If TV Shows are underrepresented, increasing investment in original series could improve long-term user engagement and retention.

### Recommendation
- Continue expanding high-performing movie content while evaluating opportunities to increase investment in TV Shows if user demand supports it.
- Analyze viewer engagement separately to determine whether movies or TV shows drive higher watch time.
